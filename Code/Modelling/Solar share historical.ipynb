Solar Share Data — Save to Data/Cleaned
========================================
Saves the historical Italian NORD zone solar share of generation
as a cleaned parquet file for use across all notebooks.
 
Source: GSE / Terna annual statistics
Reference: 11.5% (2024) → penetration_index = solar_share / 11.5
"""

```python
import pandas as pd
from pathlib import Path
```

```python
# ── paths ─────────────────────────────────────────────────────────────────────
output_path = Path("../../Data/Cleaned")
output_path.mkdir(parents=True, exist_ok=True)
```

```python
# ── solar share data ──────────────────────────────────────────────────────────
REFERENCE_SHARE = 11.5   # 2024 reference for penetration index normalisation
 
solar_share_pct = {
    2005:  0.02,
    2006:  0.03,
    2007:  0.05,
    2008:  0.10,
    2009:  0.26,
    2010:  0.63,
    2011:  3.00,
    2012:  5.70,
    2013:  7.40,
    2014:  7.90,
    2015:  7.80,
    2016:  7.90,
    2017:  7.70,
    2018:  7.90,
    2019:  8.60,
    2020:  9.50,
    2021:  9.50,
    2022:  9.50,
    2023: 12.30,
    2024: 11.50,
    2025: 13.00
}
```

```python
# ── build dataframe ───────────────────────────────────────────────────────────
df_solar_share = pd.DataFrame({
    "year":              list(solar_share_pct.keys()),
    "solar_share_pct":   list(solar_share_pct.values()),
})
 
df_solar_share["penetration_index"] = (
    df_solar_share["solar_share_pct"] / REFERENCE_SHARE
)
df_solar_share["reference_share_pct"] = REFERENCE_SHARE
df_solar_share["source"]             = "GSE/Terna"
df_solar_share["zone"]               = "NORD"
df_solar_share["notes"]              = (
    "penetration_index = solar_share_pct / reference_share_pct  "
    "(pen=1.0 at 2024)"
)
```

```python
# ── save ──────────────────────────────────────────────────────────────────────
out_file = output_path / "df_solar_share.parquet"
df_solar_share.to_parquet(out_file, index=False)
 
print("── Solar share data saved ────────────────────────────────────────")
print(f"  Path  : {out_file.resolve()}")
print(f"  Rows  : {len(df_solar_share)}")
print(f"  Cols  : {list(df_solar_share.columns)}")
print(f"\n  Reference share : {REFERENCE_SHARE}% (2024 = pen=1.0)")
print(f"\n  Preview:")
print(df_solar_share[["year","solar_share_pct","penetration_index"]]
      .to_string(index=False))
```
